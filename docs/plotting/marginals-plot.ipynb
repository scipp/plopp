# Marginals plot

The marginals plot shows a 2-dimensional image together with the 1-dimensional lines obtained from summing over each dimension.

```python
import plopp as pp
```

## Basic marginals plot



```python
da = pp.data.data2d()
pp.marginals_plot(da)
```

## Control the side of the vertical plot

Above, the 1D plot as a function of `y` is placed on the left and its data axis increases from right to left.
We can alternatively show it on the right. This also results in the axis increasing from left to right as is the more common case:

```python
pp.marginals_plot(da, side="right")
```

## Extra arguments

`marginals_plot` supports some additional arguments like `title`. It also forward any extra arguments to the underlying plot functions, in this example, `grid`. Note that the arguments are passed to *all* three subplots.

```python
pp.marginals_plot(da, title="My plot", grid=True)
```
